# 🔍 Python Regular Expressions (`re`) - Complete Guide

## What are Regular Expressions?

Regular expressions (regex) are **patterns used to match character combinations** in strings.

### Import
```python
import re
```

### Key Functions Overview
| Function | Purpose |
|----------|---------|  
| `re.match()` | Match at the **beginning** of string |
| `re.search()` | Find **first match** anywhere |
| `re.findall()` | Find **all matches** as list |
| `re.finditer()` | Find all matches as **iterator** |
| `re.sub()` | **Replace** matches |
| `re.split()` | **Split** string by pattern |
| `re.compile()` | **Compile** pattern for reuse |

## 📋 Common Pattern Reference

| Pattern | Meaning | Example |
|---------|---------|---------|  
| `\d` | Digit (0-9) | `\d+` → "123" |
| `\w` | Word char (a-z, A-Z, 0-9, _) | `\w+` → "hello_123" |
| `\s` | Whitespace | `\s+` → "   " |
| `.` | Any char except newline | `a.b` → "aXb" |
| `^` | Start of string | `^Hello` |
| `$` | End of string | `world$` |
| `*` | 0 or more | `ab*` → "a", "ab", "abb" |
| `+` | 1 or more | `ab+` → "ab", "abb" |
| `?` | 0 or 1 | `ab?` → "a", "ab" |
| `{n,m}` | n to m times | `a{2,4}` → "aa", "aaa" |
| `[]` | Character set | `[aeiou]` → vowels |
| `\|` | OR | `cat\|dog` |
| `()` | Capture group | `(\d+)-(\d+)` |

---

## 🎯 Understanding Capturing Groups

The parentheses capture whatever matches inside them, letting you extract that part separately from the full match.

### How to treat them:
- **`()`** = captures and saves the matched content
- **`(?:...)`** = groups for logic but doesn't capture (non-capturing group)
- **Multiple groups** are numbered left-to-right: `group(1)`, `group(2)`, etc.

````python
# Extract email parts
pattern = r"(\w+)@(\w+\.\w+)"
match = re.search(pattern, "user@example.com")
print(match.group(1))  # "user"
print(match.group(2))  # "example.com"

# When you just need grouping logic, use non-capturing
pattern = r"(?:cats|dogs) are"  # Matches either word, but doesn't capture
````

## 🎯 Understanding `re.Match` Object

**Type:** `re.Match` object (or `None` if no match found)

### What it contains:
A `Match` object holds all information about a successful pattern match:

1. `match.group()` - The entire matched text
2. `match.groups()` - Tuple of all groups
3. `match.start()` - Start position
4. `match.end()` - End position
5. `match.span()` - Tuple of (start, end) - index
6. `match.re` - The regex pattern used
7. `match.string` - The input string

When you use `re.match()` or `re.search()`, they return a **Match object** (or `None` if no match).


````python
text = "Contact: john@example.com"
match = re.search(r"(\w+)@(\w+\.\w+)", text)

print(type(match))              # <class 're.Match'>
print(match.group(0))           # "john@example.com"
print(match.group(1))           # "john"
print(match.group(2))           # "example.com"
print(match.groups())           # ("john", "example.com")
print(match.start())            # 9
print(match.end())              # 24
print(match.span())             # (9, 24)
print(match.string)             # "Contact: john@example.com"
````

## 1️⃣ `re.match()` - Match at Beginning

````python
# re.match() only checks at the START of string
text = "Hello World"

# Match succeeds at beginning
result = re.match(r"Hello", text)
print(f"Match 'Hello': {result.group() if result else None}")

# Match fails (World not at start)
result = re.match(r"World", text)
print(f"Match 'World': {result}")

# Using .group() to get matched text
match = re.match(r"\w+", "Python123 is great")
if match:
    print(f"Matched: '{match.group()}'")
    print(f"Start: {match.start()}, End: {match.end()}")
````

## 2️⃣ `re.search()` - Find First Match Anywhere

````python
# re.search() finds first match ANYWHERE in string
text = "My phone is 123-456-7890"

# Find phone pattern
match = re.search(r"\d{3}-\d{3}-\d{4}", text)
if match:
    print(f"Phone found: {match.group()}")

# Find email in text
text2 = "Contact: john@example.com for info"
email = re.search(r"\w+@\w+\.\w+", text2)
print(f"Email: {email.group() if email else None}")
````

## 5️⃣ `re.split()` - Split by Pattern

````python
# Split by pattern
text = "apple,banana;cherry orange"

# Split by multiple delimiters
fruits = re.split(r"[,;\s]+", text)
print(f"Fruits: {fruits}")

# Split by digits
text2 = "a1b2c3d4"
parts = re.split(r"\d", text2)
print(f"Parts: {parts}")

# Limit splits
text3 = "one-two-three-four-five"
limited = re.split(r"-", text3, maxsplit=2)
print(f"Limited: {limited}")
````

## 6️⃣ Capture Groups

````python
# Capture groups with ()
date = "2024-12-25"
match = re.search(r"(\d{4})-(\d{2})-(\d{2})", date)

if match:
    print(f"Full match: {match.group(0)}")
    print(f"Year: {match.group(1)}")
    print(f"Month: {match.group(2)}")
    print(f"Day: {match.group(3)}")
    print(f"All groups: {match.groups()}")

# Named groups
pattern = r"(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})"
match = re.search(pattern, "Date: 2024-12-25")
if match:
    print(f"\nNamed - Year: {match.group('year')}")
    print(f"Dict: {match.groupdict()}")
````

## 7️⃣ Flags (Modifiers)

````python
# re.IGNORECASE (re.I) - Case insensitive
result = re.findall(r"python", "Python PYTHON python", re.I)
print(f"Case insensitive: {result}")

# re.MULTILINE (re.M) - ^ and $ match line boundaries
text = """Line 1
Line 2
Line 3"""
lines = re.findall(r"^Line \d", text, re.M)
print(f"Multiline: {lines}")

# re.DOTALL (re.S) - . matches newline too
text2 = "Hello\nWorld"
match = re.search(r"Hello.World", text2, re.S)
print(f"Dotall match: {match.group() if match else None}")

# Combine flags
result = re.findall(r"hello", "Hello\nhello\nHELLO", re.I | re.M)
print(f"Combined: {result}")
````

## 8️⃣ Real-World Patterns

````python
# Email validation
email_pattern = r"^[\w.-]+@[\w.-]+\.\w+$"
print(re.match(email_pattern, "user@domain.com") is not None)  # True
print(re.match(email_pattern, "invalid@") is not None)  # False

# URL extraction
text = "Visit https://example.com or http://test.org"
urls = re.findall(r"https?://[\w.-]+\.\w+", text)
print(f"URLs: {urls}")

# Password strength (8+ chars, 1 upper, 1 lower, 1 digit)
def validate_password(pwd):
    return bool(re.match(r"^(?=.*[a-z])(?=.*[A-Z])(?=.*\d).{8,}$", pwd))

print(f"Strong: {validate_password('Passw0rd123')}")  # True
print(f"Weak: {validate_password('password')}")  # False
````
