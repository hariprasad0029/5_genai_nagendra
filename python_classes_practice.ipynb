## 1. Basic Class Definition

**Overview:** A class is a blueprint for creating objects. It defines attributes (data) and methods (functions).

**Workflow:**
```python
class ClassName:
    # class body
    pass

obj = ClassName()  # Create instance
```

````python
# Example: Basic Class
class Dog:
    species = "Canine"  # Class attribute

# Create instances
dog1 = Dog()
dog2 = Dog()

print(f"dog1 species: {dog1.species}")
print(f"Are they same class? {type(dog1) == type(dog2)}")
````

## 2. `__init__` Constructor

**Overview:** The `__init__` method initializes object attributes when an instance is created.

**Workflow:**
```python
class ClassName:
    def __init__(self, param1, param2):
        self.attr1 = param1
        self.attr2 = param2
```

````python
# Example: Constructor
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

person1 = Person("Alice", 30)
person2 = Person("Bob", 25)

print(f"{person1.name} is {person1.age} years old")
print(f"{person2.name} is {person2.age} years old")
````

## 3. Instance vs Class Attributes

**Overview:**
- **Class attributes:** Shared by all instances
- **Instance attributes:** Unique to each instance

**Workflow:**
```python
class MyClass:
    class_attr = "shared"  # Class attribute
    
    def __init__(self, value):
        self.instance_attr = value  # Instance attribute
```

````python
# Example: Instance vs Class Attributes
class Employee:
    company = "TechCorp"  # Class attribute
    employee_count = 0
    
    def __init__(self, name, salary):
        self.name = name  # Instance attribute
        self.salary = salary
        Employee.employee_count += 1

emp1 = Employee("John", 50000)
emp2 = Employee("Jane", 60000)

print(f"Company: {Employee.company}")
print(f"Total employees: {Employee.employee_count}")
print(f"emp1 salary: {emp1.salary}, emp2 salary: {emp2.salary}")
````

## 4. Instance Methods

**Overview:** Functions defined inside a class that operate on instance data.

**Workflow:**
```python
class MyClass:
    def method_name(self, params):
        # self refers to the instance
        return self.attribute + params
```

````python
# Example: Instance Methods
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        return f"Deposited ${amount}. New balance: ${self.balance}"
    
    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient funds!"
        self.balance -= amount
        return f"Withdrew ${amount}. New balance: ${self.balance}"

account = BankAccount("Alice", 1000)
print(account.deposit(500))
print(account.withdraw(200))
````

## 5. Class Methods & Static Methods

**Overview:**
- `@classmethod`: Works with class, not instance. Uses `cls` parameter.
- `@staticmethod`: Independent utility function, no access to class/instance.

**Workflow:**
```python
@classmethod
def class_method(cls):
    return cls.class_attribute

@staticmethod
def static_method(x, y):
    return x + y
```

````python
# Example: Class and Static Methods
class Pizza:
    def __init__(self, ingredients):
        self.ingredients = ingredients
    
    @classmethod
    def margherita(cls):
        """Factory method - alternative constructor"""
        return cls(["mozzarella", "tomatoes", "basil"])
    
    @classmethod
    def pepperoni(cls):
        return cls(["mozzarella", "pepperoni", "tomatoes"])
    
    @staticmethod
    def calculate_area(radius):
        """Utility function - doesn't need class or instance"""
        import math
        return math.pi * radius ** 2

# Using class methods as factory
pizza1 = Pizza.margherita()
print(f"Margherita: {pizza1.ingredients}")

# Using static method
print(f"12-inch pizza area: {Pizza.calculate_area(6):.2f} sq inches")
````

## 6. Inheritance

**Overview:** Create new class based on existing class. Child inherits attributes/methods from parent.

**Workflow:**
```python
class Parent:
    pass

class Child(Parent):  # Child inherits from Parent
    pass
```

````python
# Example: Inheritance
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

dog = Dog("Buddy")
cat = Cat("Whiskers")

print(f"{dog.name} says: {dog.speak()}")
print(f"{cat.name} says: {cat.speak()}")
print(f"Is Dog an Animal? {isinstance(dog, Animal)}")
````

## 7. Method Overriding & super()

**Overview:** Child class can override parent methods. Use `super()` to call parent methods.

**Workflow:**
```python
class Child(Parent):
    def __init__(self, parent_attr, child_attr):
        super().__init__(parent_attr)  # Call parent's __init__
        self.child_attr = child_attr
```

````python
# Example: super() function
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    
    def get_info(self):
        return f"{self.name}: ${self.salary}"

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)  # Call parent's __init__
        self.department = department
    
    def get_info(self):
        base_info = super().get_info()  # Call parent's method
        return f"{base_info}, Department: {self.department}"

mgr = Manager("Alice", 80000, "Engineering")
print(mgr.get_info())
````

## 8. Encapsulation (Private/Protected)

**Overview:**
- `_attr`: Protected (convention, not enforced)
- `__attr`: Private (name mangling, harder to access)

**Workflow:**
```python
class MyClass:
    def __init__(self):
        self.public = "accessible"
        self._protected = "use carefully"
        self.__private = "internal use"
```

````python
# Example: Encapsulation
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private
    
    def get_balance(self):
        return self.__balance
    
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

account = BankAccount(1000)
print(f"Balance: ${account.get_balance()}")

# This won't work directly:
# print(account.__balance)  # AttributeError

# Name mangling - still accessible but discouraged:
print(f"Via mangling: ${account._BankAccount__balance}")
````

## 9. Properties & Decorators

**Overview:** `@property` creates getters/setters with attribute-like syntax.

**Workflow:**
```python
@property
def attribute(self):
    return self._attribute

@attribute.setter
def attribute(self, value):
    self._attribute = value
```

````python
# Example: Properties
class Temperature:
    def __init__(self, celsius=0):
        self._celsius = celsius
    
    @property
    def celsius(self):
        return self._celsius
    
    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature below absolute zero!")
        self._celsius = value
    
    @property
    def fahrenheit(self):
        return (self._celsius * 9/5) + 32

temp = Temperature(25)
print(f"{temp.celsius}°C = {temp.fahrenheit}°F")

temp.celsius = 100
print(f"{temp.celsius}°C = {temp.fahrenheit}°F")
````

## 10. Magic/Dunder Methods

**Overview:** Special methods like `__str__`, `__repr__`, `__len__`, `__add__` customize behavior.

**Common Methods:**
- `__str__`: Human-readable string
- `__repr__`: Developer representation
- `__len__`: len() behavior
- `__eq__`: Equality comparison
- `__add__`: + operator

````python
# Example: Magic Methods
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"Vector({self.x}, {self.y})"
    
    def __repr__(self):
        return f"Vector(x={self.x}, y={self.y})"
    
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    
    def __len__(self):
        return int((self.x**2 + self.y**2)**0.5)

v1 = Vector(3, 4)
v2 = Vector(1, 2)

print(f"v1: {v1}")
print(f"v1 + v2 = {v1 + v2}")
print(f"len(v1) = {len(v1)}")
print(f"v1 == v2: {v1 == v2}")
````

## 11. Multiple Inheritance

**Overview:** A class can inherit from multiple parent classes.

**Workflow:**
```python
class Child(Parent1, Parent2):
    pass
```

**MRO (Method Resolution Order):** Python uses C3 linearization to determine method lookup order.

````python
# Example: Multiple Inheritance
class Flyable:
    def fly(self):
        return "Flying!"

class Swimmable:
    def swim(self):
        return "Swimming!"

class Duck(Flyable, Swimmable):
    def __init__(self, name):
        self.name = name
    
    def quack(self):
        return "Quack!"

duck = Duck("Donald")
print(f"{duck.name} can: {duck.fly()}, {duck.swim()}, {duck.quack()}")

# Check MRO
print(f"\nMRO: {Duck.__mro__}")
````
