# Step-by-Step Explanation: Creating a RAG Pipeline with Gemini Embeddings and FAISS

This notebook builds a Retrieval-Augmented Generation (RAG) pipeline for Q&A over your documents. Here’s what each major step does:

1. **Document Loading**
   - Load PDF, DOCX, and TXT files from a folder using LangChain loaders.
   - Extract the raw text from each document.

2. **Text Cleaning**
   - Clean the extracted text to remove extra spaces, newlines, and non-printable characters for better downstream processing.

3. **Chunking**
   - Split the cleaned text into smaller, overlapping chunks using `RecursiveCharacterTextSplitter`.
   - This ensures each chunk fits within the context window of the embedding model and preserves context across boundaries.

4. **Embedding Generation**
   - Use Google Gemini’s embedding model (via `GoogleGenerativeAIEmbeddings`) to convert each text chunk into a high-dimensional vector (embedding).
   - These embeddings capture the semantic meaning of each chunk for later retrieval.

5. **Vector Store Creation (FAISS)**
   - Store all embeddings in a FAISS index, which enables fast similarity search (retrieval) over the chunks.
   - Save both the FAISS index and the original text chunks for later use.

**Summary:**
- You are preparing your documents for a RAG Q&A system: load → clean → chunk → embed → store.
- Later, you can retrieve the most relevant chunks for a user query and use Gemini or another LLM to answer questions based on your documents.

```python
RAG -- What is RAG 


RAG stands for Retrieval-Augmented Generation. It is a technique used in natural language processing (NLP) that combines the capabilities of retrieval-based models and generative models to improve the quality and relevance of generated text.
    


In a RAG system, a retrieval model is used to search a large corpus of documents or knowledge base to find relevant 
information based on a given input query. 
This retrieved information is then used as context for a generative model, which produces a 
response or output that is informed by the retrieved content.
The generative model can be a language model like GPT-3 or similar.
```

```python
import re

def clean_text(text):
    # Remove extra whitespace and newlines
    text = re.sub(r'\s+', ' ', text)
    # Remove non-printable/control characters
    text = re.sub(r'[\x00-\x1f\x7f-\x9f]', '', text)
    # Strip leading/trailing whitespace
    return text.strip()

# Example: Clean and preprocess all loaded document texts
cleaned_documents = []
for i, doc in enumerate(documents):
    raw_text = doc.page_content
    cleaned = clean_text(raw_text)
    cleaned_documents.append(cleaned)
    print(f"\n--- Cleaned Content of Document {i+1}: {doc.metadata.get('source', 'Unknown')} ---\n")
    print(cleaned[:1000])  # Preview first 1000 characters
```

RecursiveCharacterTextSplitter
🔍 Definition

RecursiveCharacterTextSplitter is a smarter, hierarchical version of CharacterTextSplitter.
It tries to preserve sentence and paragraph boundaries as much as possible while still respecting chunk size limits.

How It Works

It recursively attempts to split the text using a list of separators in a defined order:

["\n\n", "\n", ".", "!", "?", ",", " ", ""]


So it:

First tries to split by paragraphs (\n\n).

If chunks are still too large, splits by sentences (., !, ?).

If still too large, splits by words ( ).

If still too large, splits by characters.

This recursive process ensures semantic coherence while keeping chunks within size limits.

When to Use What

Use CharacterTextSplitter
👉 when you just need fast splitting (e.g., code files, logs, or technical text).

Use RecursiveCharacterTextSplitter
👉 when you want meaningful, readable chunks for RAG pipelines, summarization, or QA bots.

The difference in the number of chunks is due to how each splitter works:

CharacterTextSplitter splits text strictly at the specified separator (e.g., newline) and creates chunks of the given size, possibly breaking sentences or paragraphs.
RecursiveCharacterTextSplitter tries to split at more natural boundaries (like paragraphs, sentences, or words) and only falls back to character-level splitting if needed. This often results in more, but more contextually meaningful, chunks.
So, RecursiveCharacterTextSplitter usually produces more (and better-formed) chunks than CharacterTextSplitter for the same chunk size and overlap.

What Is a “JSON Character Splitter”?

When we say “JSON Character Splitter”, we mean a method of chunking large JSON files (key–value pairs, nested data, etc.) into smaller JSON-like chunks — usually by characters — so each chunk:

Fits within token limits for LLMs,

Preserves JSON structure as much as possible,

Doesn’t break key-value pairs mid-way.

So it’s similar to CharacterTextSplitter, but adapted for JSON data.


Why Regular Splitters Don’t Work Well on JSON

Normal splitters (like CharacterTextSplitter) see JSON as plain text:

{"company": "OpenAI", "products": ["GPT", "ChatGPT", "DALL·E"], "details": {"employees": 1000}}


They might split like this:

ducts": ["GPT", "ChatGPT", 
"DALL·E"], "details": {"empl
oyees": 1000}}

If your JSON is a list of objects (common in logs, datasets, etc.):

import json

def json_chunker(json_data, chunk_size=3):
    """
    Splits JSON (list of objects) into smaller JSON chunks
    each with up to `chunk_size` elements.
    """
    if not isinstance(json_data, list):
        raise ValueError("Input JSON must be a list of objects.")

    for i in range(0, len(json_data), chunk_size):
        yield json_data[i:i + chunk_size]

# Example JSON
data = [
    {"id": 1, "name": "Alpha"},
    {"id": 2, "name": "Beta"},
    {"id": 3, "name": "Gamma"},
    {"id": 4, "name": "Delta"},
    {"id": 5, "name": "Epsilon"}
]

for i, chunk in enumerate(json_chunker(data, chunk_size=2)):
    print(f"\n--- Chunk {i+1} ---")
    print(json.dumps(chunk, indent=2))

why text-embedding-3-small vs text-embedding-3-large?
the difference between text-embedding-3-small and text-embedding-3-large mainly comes down to vector size, accuracy, cost, and speed. Let’s break it down clearly 👇

1. Model Overview

| Model                      | Dimensions | Accuracy | Cost                                      | Typical Use                                                                                  |
| -------------------------- | ---------- | -------- | ----------------------------------------- | -------------------------------------------------------------------------------------------- |
| **text-embedding-3-small** | **1,536**  | Medium   | 💲 **Cheaper (~1/5th the cost of large)** | Fast semantic search, clustering, lightweight retrieval                                      |
| **text-embedding-3-large** | **3,072**  | High     | 💰 **More expensive**                     | RAG (Retrieval Augmented Generation), fine-grained semantic tasks, reasoning-heavy retrieval |


2. Core Differences
🔹 (a) Embedding Quality

3-large produces richer, more semantically accurate embeddings — it captures deeper relationships between words and phrases.

3-small works well for most “search and clustering” tasks, but may miss subtle context or long-range meaning.

👉 Example:

3-small might think “car” and “truck” are closer than “car” and “electric vehicle.”

3-large would understand the latter pair are more conceptually related.

(b) Dimensionality

3-large: 3,072 floats → more expressive but doubles storage & compute.

3-small: 1,536 floats → smaller, faster for retrieval and similarity search, but less expressive.

(c) Cost & Speed
| Model       | Cost per 1K tokens | Speed                         |
| ----------- | ------------------ | ----------------------------- |
| **3-small** | ~$0.02             | ⚡ Faster                      |
| **3-large** | ~$0.13             | 🐢 Slower (but more accurate) |

(d) Use Case Fit

| Use Case                                             | Recommended Model |
| ---------------------------------------------------- | ----------------- |
| Document retrieval (RAG) needing precise context     | ✅ `3-large`       |
| FAQ search, semantic search on large corpus          | ✅ `3-small`       |
| Topic clustering or deduplication                    | ✅ `3-small`       |
| Long context reasoning or knowledge graph embeddings | ✅ `3-large`       |

- **`faiss_gemini_docs.pkl`**:
    - This file contains the list of all text chunks (strings) that were created from your documents using the recursive character text splitter.
    - It is saved using Python's `pickle` module, which serializes the list of chunks so you can reload them later.
    - When you perform a similarity search, you use this file to map the retrieved embedding vectors back to their original text content.

- **`faiss_gemini_index.idx`**:
    - This file is the FAISS vector index, which stores all the embeddings (vectors) generated from your text chunks.
    - FAISS (Facebook AI Similarity Search) is a library for efficient similarity search and clustering of dense vectors.
    - This index allows you to quickly find the most similar chunks to a user query by comparing their embeddings.
    - You load this file to perform fast vector searches without recomputing embeddings or rebuilding the index.

**In summary:**
- `faiss_gemini_docs.pkl` = the actual text chunks.
- `faiss_gemini_index.idx` = the vector representations (embeddings) of those chunks, stored for fast retrieval.

What is a Vector Database (recap)

A vector database stores and retrieves embedding vectors (numerical representations of text, images, etc.) efficiently using approximate nearest neighbor (ANN) search.
This enables semantic search — instead of keyword matching, the DB finds conceptually similar documents.

   - Highly efficient for large-scale similarity search.
   - Supports various indexing methods (e.g., IVF, HNSW).
   - Best for on-premise or self-hosted solutions.

```python
# 1. Load the FAISS index and text chunks
import faiss
import pickle
import numpy as np

def load_faiss_and_chunks(index_path="faiss_gemini_index.idx", chunks_path="faiss_gemini_docs.pkl"):
    faiss_index = faiss.read_index(index_path)
    with open(chunks_path, "rb") as f:
        chunks = pickle.load(f)
    return faiss_index, chunks
```

```python
# 2. Embed the user query with Gemini
from langchain_google_genai import GoogleGenerativeAIEmbeddings

def embed_query(query, api_key):
    embedding_model = GoogleGenerativeAIEmbeddings(model="models/embedding-001", google_api_key=api_key)
    query_embedding = embedding_model.embed_query(query)
    return np.array(query_embedding).astype('float32')
```

```python
# 3. Use FAISS to retrieve the most similar chunks
def retrieve_top_k_chunks(query_embedding, faiss_index, chunks, k=3):
    D, I = faiss_index.search(query_embedding.reshape(1, -1), k)
    retrieved_chunks = [chunks[i] for i in I[0]]
    return retrieved_chunks
```

```python
# 4. Build a context passage from those chunks
def build_context_passage(chunks):
    return "\n\n".join(chunks)
```

```python
# 5. Pass the context and question to an LLM (Gemini) to generate an answer
import requests
# Example of prompt engineering for better answers
def build_prompt(context, question):
    prompt = (
        "You are automotive expert assistant. Use ONLY the provided context to answer the question. "
        "If the answer is not in the context, say 'I don't know based on the provided information.'\n\n"
        f"Context:\n{context}\n\n"
        f"Question: {question}\n"
        "Answer (be concise and reference the context): strictly idf the use question is not part of embeddings, just give answer as Please qask from domain knowledge only"
    )
    return prompt

def generate_answer_with_prompt_engineering(context, question, api_key):
    url = "https://generativelanguage.googleapis.com/v1beta/models/gemini-pro:generateContent"
    headers = {"Content-Type": "application/json"}
    prompt = build_prompt(context, question)
    data = {
        "contents": [{"parts": [{"text": prompt}]}]
    }
    response = requests.post(f"{url}?key={api_key}", headers=headers, json=data)
    if response.status_code == 200:
        return response.json()['candidates'][0]['content']['parts'][0]['text']
    else:
        print("Error:", response.text)
        return None
```

When explaining LLM generation parameters like temperature, maxOutputTokens, topP, and topK in an interview, you want to be clear, concise, and give examples so even a non-technical interviewer can follow. Here’s a structured way to explain them:



1️⃣ temperature

What it is:

Controls creativity / randomness of the model’s output.

Higher → more creative/random responses.

Lower → more deterministic/predictable responses.

Analogy:

Think of it like a thermostat for creativity.

temperature=0 → “robotic” and precise answers.

temperature=1 → “imaginative” answers.

Example:

Question: “Write a short story about a cat.”

temperature=0 → “A cat sat on the mat and slept.”

temperature=0.8 → “A mischievous tabby cat danced across rooftops under a golden sunset.”

Interview Tip:

Emphasize: “I can control creativity vs predictability using temperature.”


2️⃣ maxOutputTokens

What it is:

Maximum number of tokens the model will generate.

One token ≈ 4 characters on average (varies by language).

Analogy:

Think of it as a word limit in an essay or paragraph.

Example:

maxOutputTokens=50 → response will be limited to about 50 tokens.

Interview Tip:

Mention: “It prevents the model from generating overly long or irrelevant responses, useful for efficiency and cost control.”


3️⃣ topP (nucleus sampling)

What it is:

Controls probabilistic sampling by keeping the most likely tokens that make up P% of total probability.

Helps the model pick from a subset of the most probable next words.

Analogy:

Imagine a bag of words sorted by likelihood. TopP=0.9 → pick words from the top 90% probability mass.

Example:

High topP (0.9–1.0) → more creative, diverse outputs.

Low topP (0.3–0.5) → more conservative, focused output.

Interview Tip:

Say: “topP is a way to balance randomness and focus — more controlled than temperature in some cases.”


4️⃣ topK

What it is:

Limits the next token selection to the top K most probable tokens.

Only considers the K most likely words, ignores the rest.

Analogy:

Like having a shortlist of K options to choose from for the next word.

Example:

topK=5 → model picks next word from only the 5 most probable words.

Interview Tip:

Say: “topK ensures the model doesn’t pick very unlikely words, keeping responses relevant but still slightly diverse.”

Quick Comparison Table (for interviewer clarity)

| Parameter           | Purpose                 | Effect                                        | Example                                  |
| ------------------- | ----------------------- | --------------------------------------------- | ---------------------------------------- |
| **temperature**     | Randomness / creativity | Higher = more creative, lower = deterministic | 0 → factual answer, 0.8 → creative story |
| **maxOutputTokens** | Output length limit     | Limits response length                        | 50 tokens max                            |
| **topP**            | Probabilistic sampling  | Keeps top P% probability mass                 | 0.9 → pick from most likely 90% tokens   |
| **topK**            | Token shortlist         | Picks next word from top K options            | 5 → only 5 most likely tokens            |

```python
# --- Evaluation Metrics for RAG Q&A ---
# Install required packages if not already installed
# !pip install scikit-learn rouge-score nltk --quiet

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from rouge_score import rouge_scorer
import nltk
# Download punkt and punkt sentence models if not already present
nltk.download('punkt', quiet=True)
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

# Accuracy: Percentage of answers that are fully correct (exact match)
def compute_accuracy(expected, predicted):
    return accuracy_score(expected, predicted)

# Precision, Recall, F1: For multi-label or extractive answers (expects lists of labels)
def compute_prf1(expected, predicted):
    return {
        'precision': precision_score(expected, predicted, average='micro', zero_division=0),
        'recall': recall_score(expected, predicted, average='micro', zero_division=0),
        'f1': f1_score(expected, predicted, average='micro', zero_division=0)
    }

# BLEU Score: For comparing generated answers to reference answers (token overlap)
def compute_bleu(reference, candidate):
    try:
        smoothie = SmoothingFunction().method4
        # Use sent_tokenize to ensure punkt sentence model is loaded
        _ = sent_tokenize(reference)
        ref_tokens = word_tokenize(reference)
        cand_tokens = word_tokenize(candidate)
        return sentence_bleu([ref_tokens], cand_tokens, smoothing_function=smoothie)
    except LookupError as e:
        print("NLTK resource missing. Downloading 'punkt'...")
        nltk.download('punkt')
        _ = sent_tokenize(reference)
        ref_tokens = word_tokenize(reference)
        cand_tokens = word_tokenize(candidate)
        return sentence_bleu([ref_tokens], cand_tokens, smoothing_function=smoothie)

# ROUGE Score: For comparing generated answers to reference answers (recall-oriented)
def compute_rouge(reference, candidate):
    scorer = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=True)
    return scorer.score(reference, candidate)['rougeL'].fmeasure

# Example usage:
expected = ["Battery Management System", "Centralized, Distributed, Modular", "I don't know"]
predicted = ["Battery Management System", "Centralized, Distributed, Modular", "I don't know"]
print('Accuracy:', compute_accuracy(expected, predicted))
print('BLEU:', compute_bleu(expected[0], predicted[0]))
print('ROUGE:', compute_rouge(expected[0], predicted[0]))
```
